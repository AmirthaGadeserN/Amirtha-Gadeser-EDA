```python
import numpy as np
import pandas as pd
data = np.arange(15, 30).reshape(5, 3)

dfx = pd.DataFrame(
    data,
    index=['apple', 'banana', 'kiwi', 'grapes', 'mango'],
    columns=['store1', 'store2', 'store3']
)

dfx['store4'] = np.nan
dfx.loc['watermelon'] = np.arange(15, 19)
dfx.loc['oranges'] = np.nan
dfx['store5'] = np.nan
dfx.loc['apple', 'store4'] = 20

print("Original DataFrame:")
print(dfx)
print("Missing values:")
print(dfx.isnull())

print("\nNon-missing values:")
print(dfx.notnull())
print("Missing values in each column:")
print(dfx.isnull().sum())

print("\nTotal missing values:")
print(dfx.isnull().sum().sum())

print("\nNumber of non-null values:")
print(dfx.count())
print("Non-null values in store4:")
print(dfx.store4[dfx.store4.notnull()])

print("\nstore4 after dropping NaN:")
print(dfx.store4.dropna())

print("\nDrop rows with NaN:")
print(dfx.dropna())

print("\nDrop rows where all values are NaN:")
print(dfx.dropna(how='all'))

print("\nDrop columns where all values are NaN:")
print(dfx.dropna(how='all', axis=1))

print("\nColumns with at least 5 non-null values:")
print(dfx.dropna(thresh=5, axis=1))
ar1 = np.array([100, 200, np.nan, 300])
ser1 = pd.Series(ar1)

print("NumPy mean:", ar1.mean())
print("Pandas mean:", ser1.mean())

ser2 = dfx.store4

print("\nSum of store4:")
print(ser2.sum())

print("\nMean of store4:")
print(ser2.mean())

print("\nCumulative sum of store4:")
print(ser2.cumsum())
filledDf = dfx.fillna(0)

print("DataFrame after filling NaN with 0:")
print(filledDf)

print("\nMean before filling:")
print(dfx.mean(numeric_only=True))

print("\nMean after filling:")
print(filledDf.mean(numeric_only=True))
print("Forward fill:")
print(dfx.store4.ffill())

print("\nBackward fill:")
print(dfx.store4.bfill())
```
