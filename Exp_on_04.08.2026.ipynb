```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import KBinsDiscretizer, Binarizer

file_path = "california_housing_test.csv"

try:
    df = pd.read_csv(file_path)
except FileNotFoundError:
    df = pd.read_csv("sample_data/california_housing_test.csv")

print("--- Original Dataset Sample ---")
print(df[['housing_median_age', 'median_income', 'median_house_value']].head())
print("\n" + "="*80 + "\n")

df['age_equal_width'] = pd.cut(
    df['housing_median_age'],
    bins=4,
    labels=['Very New', 'Moderate', 'Old', 'Very Old']
)

df['income_quantile_bins'] = pd.qcut(
    df['median_income'],
    q=4,
    labels=['Low', 'Medium', 'High', 'Very High']
)

custom_bins = [0, 100000, 200000, 350000, np.inf]
custom_labels = ['Affordable', 'Moderate', 'Expensive', 'Luxury']

df['house_value_custom'] = pd.cut(
    df['median_house_value'],
    bins=custom_bins,
    labels=custom_labels
)

binarizer = Binarizer(threshold=5.0)
df['income_high_flag'] = binarizer.fit_transform(df[['median_income']])

kbd_kmeans = KBinsDiscretizer(n_bins=3, encode='ordinal', strategy='kmeans', random_state=42)
df['house_value_kmeans_bin'] = kbd_kmeans.fit_transform(df[['median_house_value']])

discretized_table = df[[
    'housing_median_age', 'age_equal_width',
    'median_income', 'income_quantile_bins', 'income_high_flag',
    'median_house_value', 'house_value_custom', 'house_value_kmeans_bin'
]]

print("--- Transformed Table with Discretized Features ---")
print(discretized_table.head(10))
print("\n" + "="*80 + "\n")

print("--- Bin Distribution Counts ---")
print("\n[Equal-Width] housing_median_age:")
print(df['age_equal_width'].value_counts())

print("\n[Equal-Frequency / Quantiles] median_income:")
print(df['income_quantile_bins'].value_counts())

print("\n[Custom Intervals] median_house_value:")
print(df['house_value_custom'].value_counts())

print("\n[Binarization Flag (>5.0)] median_income:")
print(df['income_high_flag'].value_counts())
```
