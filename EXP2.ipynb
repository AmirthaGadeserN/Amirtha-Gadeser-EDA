```python
import pandas as pd

df = pd.read_csv("happiness.csv")

print("1. First 5 Rows:")
print(df.head())

print("\n2. Last 5 Rows:")
print(df.tail())

print("\n3. Dataset Information:")
print(df.info())

print("\n4. Shape of Dataset:")
print(df.shape)

print("\n5. Column Names:")
print(df.columns)

print("\n6. Summary Statistics:")
print(df.describe())

print("\n7. Missing Values:")
print(df.isnull().sum())

print("\n8. Happiness Counts:")
print(df["happy"].value_counts())

print("\n9. Top 10 Highest Income Records:")
print(df.sort_values(by="income", ascending=False).head(10))

print("\n10. Data Types")
print(df.dtypes)
```
