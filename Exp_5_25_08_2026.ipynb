```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = sns.load_dataset('iris')

print("--- First 5 Rows ---")
print(df.head())
print("\n--- Dataset Summary ---")
print(df.info())
print("\n--- Statistical Overview ---")
print(df.describe())
print("\n--- Class Distribution ---")
print(df['species'].value_counts())

sns.pairplot(df, hue='species', palette='Set2')
plt.suptitle("Pairplot of Iris Features by Species", y=1.02)
plt.show()
plt.figure(figsize=(10, 6))
df_melted = pd.melt(df, id_vars='species', var_name='Feature', value_name='Value')
sns.boxplot(x='Feature', y='Value', hue='species', data=df_melted, palette='Set2')
plt.title("Feature Distribution Across Species")
plt.show()

plt.figure(figsize=(8, 6))
sns.heatmap(df.drop(columns=['species']).corr(), annot=True, cmap='coolwarm', fmt=".2f")
plt.title("Correlation Matrix of Iris Features")
plt.show()
```
