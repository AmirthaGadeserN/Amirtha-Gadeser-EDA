```
#univariate analysis
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv("WineQT.csv")

print("--- Dataset Overview ---")
print(f"Dataset Shape: {df.shape}\n")

print("--- Summary Statistics for Numerical Features ---")

print(df.describe().T)
print("\n--- Missing Values Check ---")
print(df.isnull().sum())

plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='quality', palette='muted')
plt.title('Distribution of Wine Quality (Target Variable)', fontsize=14)
plt.xlabel('Quality Rating')
plt.ylabel('Count')
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.histplot(data=df, x='alcohol', kde=True, ax=axes[0], color='crimson')
axes[0].set_title('Distribution of Alcohol Content', fontsize=12)
axes[0].set_xlabel('Alcohol (%)')

sns.histplot(data=df, x='pH', kde=True, ax=axes[1], color='darkblue')
axes[1].set_title('Distribution of pH Levels', fontsize=12)
axes[1].set_xlabel('pH Value')

plt.tight_layout()
plt.show()

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

sns.boxplot(data=df, x='residual sugar', ax=axes[0], color='gold')
axes[0].set_title('Box Plot of Residual Sugar (Outlier Detection)', fontsize=12)

sns.boxplot(data=df, x='sulphates', ax=axes[1], color='teal')
axes[1].set_title('Box Plot of Sulphates (Outlier Detection)', fontsize=12)

plt.tight_layout()
plt.show()

features_to_plot = df.drop(columns=['Id', 'quality'], errors='ignore').columns

print("\nGenerating histograms for all continuous variables...")
df[features_to_plot].hist(figsize=(15, 12), bins=20, edgecolor='black')
plt.suptitle('Univariate Histograms of All Wine Chemical Features', fontsize=16, y=1.02)
plt.tight_layout()
plt.show()
```

```
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

df = pd.read_csv("WineQT.csv")

print(df.head())
print("Columns:", df.columns.tolist())

plt.figure(figsize=(12, 8))
correlation_matrix = df.corr()
sns.heatmap(
    correlation_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5
)
plt.title("Correlation Heatmap - Bivariate Analysis")
plt.show()

plt.figure(figsize=(8, 6))
sns.scatterplot(
    x="alcohol", y="density", data=df, alpha=0.6, color="purple"
)
plt.title("Scatter Plot of Alcohol vs Density")
plt.xlabel("Alcohol")
plt.ylabel("Density")
plt.show()

plt.figure(figsize=(8, 6))
sns.boxplot(x="quality", y="alcohol", data=df, palette="Set2")
plt.title("Box Plot of Alcohol Content by Wine Quality")
plt.xlabel("Quality")
plt.ylabel("Alcohol")
plt.show()
```

```
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

df = pd.read_csv("WineQT.csv")

print("Dataset Shape:", df.shape)
print(df.head())

plt.figure(figsize=(10, 8))
corr_matrix = df.corr()
sns.heatmap(
    corr_matrix, annot=True, cmap="coolwarm", fmt=".2f", linewidths=0.5
)
plt.title("Correlation Heatmap of Wine Physicochemical Features")
plt.tight_layout()
plt.show()

subset_features = ["fixed acidity", "volatile acidity", "alcohol", "quality"]
sns.pairplot(df[subset_features], hue="quality", palette="viridis")
plt.suptitle("Pairplot of Key Wine Features by Quality", y=1.02)
plt.show()

X = df.drop(
    columns=[col for col in ["quality", "Id"] if col in df.columns],
    errors="ignore",
)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

pca = PCA(n_components=2)
principal_components = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame(
    data=principal_components, columns=["Principal Component 1", "Principal Component 2"]
)
pca_df["quality"] = df["quality"].values

plt.figure(figsize=(8, 6))
sns.scatterplot(
    x="Principal Component 1",
    y="Principal Component 2",
    hue="quality",
    data=pca_df,
    palette="Set1",
    alpha=0.7,
)
plt.title("PCA of Wine Dataset (Colored by Quality)")
plt.xlabel(f"PC1 ({pca.explained_variance_ratio_[0]*100:.2f}% variance)")
plt.ylabel(f"PC2 ({pca.explained_variance_ratio_[1]*100:.2f}% variance)")
plt.legend(title="Quality", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
```
